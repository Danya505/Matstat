## f) Сгенерируйте выборку объема n = 100 для некоторого параметра Θ. Вычислите указанные выше доверительные интервалы для доверительной ввероятности 0.95.

```python
import numpy as np
import random as rd

theta = 2

def p(x:float) -> float:
    if 2 * theta >= x >= theta:
        return 1/x
    return 0
def F(x:float) -> float:
    if 2 * theta >= x >= theta:
        return x/theta - 1
    elif x >= 2 * theta:
        return 1
    return 0
def F_inv(y:float) -> float:
    return theta * (y + 1)

N = 100
selection = []

for _ in range(N):
    y = rd.random()
    selection.append(F_inv(y))

beta = 0.95
x_max = np.max(selection)

lower_limit = x_max/(1 + ((1 + beta) / 2) ** (1 / N))
upper_limit = x_max/(1 + ((1 - beta) / 2) ** (1 / N))

print(f"Доверительный интервал : ({lower_limit}, {upper_limit}), длина : {upper_limit - lower_limit}")
```

```python
theta_1 = 2 * np.mean(selection) / 3
s = np.sqrt((np.sum([(x - np.mean(selection)) ** 2 for x in selection]))/(N-1))

lower_limit = theta_1 - 2 * s * 1.96 * np.sqrt(N - 1)/(3 * N)
upper_limit = theta_1 + 2 * s * 1.96 * np.sqrt(N - 1)/(3 * N)

print(f"Доверительный интервал : ({lower_limit}, {upper_limit}), длина : {upper_limit - lower_limit}")
```

## g) Численно постройте бутстраповский доверительный интервал.

```python
h_wave = (N + 1) * x_max / (2 * N + 1)

bootstrap_delta = []
for _ in range(1000):
    sample = np.random.choice(selection, size=N)
    bootstrap_delta.append((N + 1) * np.max(sample) / (2 * N + 1) - h_wave)

sorted_bootstrap_delta = np.sort(bootstrap_delta)
K1 = sorted_bootstrap_delta[int(((1 - beta) * 1000) / 2)]
K2 = sorted_bootstrap_delta[int(((1 + beta) * 1000) / 2)]

lower_bound = h_wave - K2
upper_bound = h_wave - K1

print(f"Доверительный интервал: ({lower_bound}, {upper_bound}), длина: {upper_bound - lower_bound}")
```
