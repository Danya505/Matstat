## d) Сгенерируйте выборку n = 100 для некоторого значения параметра Θ. Вычислите указанные выше доверительные интервалы для доверительной вероятности 0.95.

```python
import numpy as np
import random as rd

theta = 2

def p(x:float) -> float:
    if x >= 1:
        return (theta - 1)/x ** theta
    return 0
def F(x:float) -> float:
    if x >= 1:
        return 1 - x ** (1 - theta)
    return 0
def F_inv(y: float, thet: float) -> float:
    return (1 - y) ** (1 / (1 - thet))

N = 100
selection = []

for _ in range(N):
    y = rd.random()
    selection.append(F_inv(y, theta))

beta = 0.95

mean_ln = np.mean([np.log(x) for x in selection])
g_theta = 2 ** (mean_ln)

delta = 1.96 * g_theta * np.log(2) * mean_ln / np.sqrt(N)

lower_limit = g_theta - delta
upper_limit = g_theta + delta

# сравнение с реальной медианой
variation_row = sorted(selection)

print(f"Доверительный интервал: ({lower_limit}, {upper_limit}), длина: {upper_limit - lower_limit}")
print("Медиана: ", (variation_row[49] + variation_row[50])/2)
```

```python
# Асимптотический доверительный интервал

lower_limit = 1 - (1.96 - np.sqrt(N)) / (mean_ln * np.sqrt(N))
upper_limit = 1 + (1.96 + np.sqrt(N)) / (mean_ln * np.sqrt(N))

print(f"Доверительный интервал: ({lower_limit}, {upper_limit}), длина: {upper_limit - lower_limit}")
```

## e) Численно постройте бутстраповский доверительный интервал двумя способами, используя параметрический бутстрап и непараметрический бутстрап.

```python
# Непараметрический бутстрап
h_wave = 1 + 1 / mean_ln

bootstrap_delta = []
for _ in range(1000):
    sample = np.random.choice(selection, size=N)
    bootstrap_delta.append(1 + 1 / (np.mean([np.log(x) for x in sample])) - h_wave)

sorted_bootstrap_delta = np.sort(bootstrap_delta)
t1 = sorted_bootstrap_delta[int(((1 - beta) * 1000) / 2)]
t2 = sorted_bootstrap_delta[int(((1 + beta) * 1000) / 2)]

lower_limit = h_wave - t2
upper_limit = h_wave - t1

print(f"Доверительный интервал: ({lower_limit}, {upper_limit}), длина: {upper_limit - lower_limit}")
```

```python
# Параметрический бутстрап
iter = 50000

h_wave = 1 + 1 / mean_ln
bootstrap_delta = []

for _ in range(iter):
    sample = []
    for _ in range(100):
        y = rd.random()
        sample.append(F_inv(y, h_wave))
    bootstrap_delta.append(1 + 1 / (np.mean([np.log(x) for x in sample])) - h_wave)

sorted_bootstrap_delta = np.sort(bootstrap_delta)
t1 = sorted_bootstrap_delta[int(((1 - beta) * iter) / 2)]
t2 = sorted_bootstrap_delta[int(((1 + beta) * iter) / 2)]

lower_limit = h_wave - t2
upper_limit = h_wave - t1

print(f"Доверительный интервал: ({lower_limit}, {upper_limit}), длина: {upper_limit - lower_limit}")
```
